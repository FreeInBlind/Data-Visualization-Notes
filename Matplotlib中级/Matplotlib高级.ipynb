### 1 图例

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0,2*np.pi,50)
plt.figure(figsize=(9,6))
y1 = np.sin(x)
y2 = np.cos(x)
plt.plot(x,y1,label = 'Sin')
plt.plot(x,y2,label = 'Cos')
plt.legend(loc='center',
           fontsize=15,
           ncol=2,
           bbox_to_anchor=[0.5,1.1,0,0]) #显示图例
```

### 2 脊柱移动

```python
plt.figure(figsize=(9,6))

plt.plot(x,y1,x,y2)

ax = plt.gca() #获取当前视图Axes

#右边和上边的脊柱消失
ax.spines['right'].set_color('white')
ax.spines['top'].set_color('white') #.set_alpha(0)也可以

#移动
ax.spines['bottom'].set_position(('data',0))
ax.spines['left'].set_position(('data',np.pi))
plt.yticks([-1,0,1])
xticks = np.arange(0,2*np.pi+0.001,np.pi/2)
_ = plt.xticks(xticks,labels = [r'$-\pi$',r'$-\frac{\pi}{2}$',0,r'$\frac{\pi}{2}$',r'$\pi$'])
```

### 3 图片保存

```python
fig = plt.figure(linewidth=4,facecolor='y',edgecolor='b') #限宽,底色，边框色
fig.set_figwidth(9)
fig.set_figheight(6)
plt.plot(x,y1,x,y2)

ax = plt.gca()
ax.set_facecolor('lightgreen')

plt.legend(['Sin','Cos'],loc='center',bbox_to_anchor=[0.5,1.1,0,0],ncol=2,fontsize=20)
plt.title('Sin vs Cos',pad=60,fontsize=30)
plt.ylabel('f(x) = sin(x) vs cos(x)',rotation=20,fontsize=20,labelpad=100)
plt.savefig('./1.png',dpi=100,bbox_inches='tight') #紧凑型尺寸，避免图片丢失数据
```

### 4 风格和样式

```python
x = np.linspace(0,2*np.pi,20)
y1 = np.sin(x)
y2 = np.cos(x)

#设置颜色，线型，点型
plt.plot(x,y1,color='indigo',ls='-',marker='p')
plt.plot(x,y2,color='#FF00EE',ls='--',marker='o')
plt.plot(x,y1+y2,color=(0.2,0.7,0.2),marker='*',ls=':')
plt.plot(x,y1+y2,color=np.random.random(3),marker='*',ls=':')
plt.plot(x,y1+2*y2,linewidth=3,alpha=0.7,color='orange') #线宽、透明度
plt.plot(x,2*y1-y2,'bo--') #参数连用(color='b',marker='o',ls='--')
```

```python
line, = plt.plot(x,y1,color=np.random.random(3),ls='-.')
line.set_dashes([5,3,1,2]) #控制该线的间隔，即自定义linestyle
```

```python
#更多属性设置
def f(x):
    return np.exp(-x)*np.cos(2*np.pi*x)
x = np.linspace(0,5,50)
plt.figure(figsize=(9,6))
plt.plot(x,f(x),color='purple',marker='o',ls='--',lw=2,alpha=0.6,
         markerfacecolor='red', #点颜色
         markersize=10, #点大小
         markeredgecolor='green', #点边缘颜色
         markeredgewidth=3) #点边缘宽度
plt.xticks(size=18)
_=plt.yticks(size=18)
```

### 6 嵌套

```python
x = np.linspace(-np.pi,np.pi,25)
y = np.sin(x)
fig = plt.figure(figsize=(9,6))

plt.plot(x,y)

#嵌套方式一：axes轴域（横纵坐标范围），子视图
ax = plt.axes([0.2,0.55,0.3,0.3]) #参数范围[left,bottom,width,height]
ax.plot(x,y,color='g')

#嵌套方式二
ax = fig.add_axes([0.55,0.2,0.3,0.3]) #使用视图对象添加子视图
ax.plot(x,y,color='r')
```

### 7 双轴布局

```python
x = np.linspace(-np.pi,np.pi,100)
y1 = np.sin(x)
y2 = np.exp(x)

plt.figure(figsize=(9,6))
plt.rcParams['font.size'] = 20

ax1 = plt.gca()
ax1.plot(x,y2,color='green')
ax1.set_xlabel('time(s)')
ax1.set_ylabel('exp',color='green')

ax2 = ax1.twinx() #共用x轴
ax2.plot(x,y1,color = 'red')
ax2.set_ylabel('sin',color='red')
plt.tight_layout() #紧凑布局
plt.savefig('./双轴布局.png',dpi=100)
```
