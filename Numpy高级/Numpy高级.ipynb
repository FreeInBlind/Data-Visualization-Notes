### 1 数据类型

```python
import numpy as np

arr1 = np.array([1,2,3,127,126],dtype = np.int8)
display(arr1,2**8) #256 数字范围：-123~0~127

arr2 = np.array([1,2,5,8,255,253],dtype = np.uint8) #无符号的8位表示数字（无负号）
display(arr2) #0~255
```

```python
arr1 = np.random.randint(0,150,size = (10000,3),dtype = np.int16)
arr2 = arr1.astype(np.int64) #原数据的基础上，对数据类型进行调整
display(arr1,arr2)
np.save('arr1',arr1)
np.save('arr2',arr2)
```

```python
arr3 = np.random.randn(100)
display(arr3,arr3.dtype)
arr4 = arr3.astype(np.float16)
display(arr4)
```

### 3 复制和视图

#### 3.1 完全没有复制

```python
import numpy as np
a = np.random.randint(0,100,size = (4,5))
b = a
print(a is b) #返回True，a和b是两个不同名字对应同一个内存对象
b[0,0] = 1024 #命运共同体
display(a,b)
```

#### 3.2 查看或浅拷贝

```python
import numpy as np
a = np.random.randint(0,100,size = (4,5))
b = a.view()
print(a is b)
print(b.base is a)
print(b.flags.owndata)
print(a.flags.owndata)
b[0,0] = 1024
display(a,b)
```

#### 3.3 深拷贝

```python
import numpy as np
a = np.random.randint(0,100,size = (4,5))
b = a.copy()
print(b is a)
print(b.base is a)
print(b.flags.owndata)
print(a.flags.owndata)
b[0,0] = 1024
display(a,b)
```

### 4 基本的索引和切片

#### 4.1 深拷贝的应用

```python
import numpy as np
import gc #垃圾回收
a = np.arange(1e8)
display(a)
b = a[::1000000].copy()
del a #不再需要a，删除大内存的a
gc.collect()
display(b,b.shape) #shape(100,)
```
