#### 1.1 图形绘制

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,2*np.pi,50)
y = np.sin(x)
plt.figure(figsize = (9,6)) #图形尺寸
plt.plot(x,y,color = 'green')
plt.plot(x,np.cos(x),color = 'm')
#linestyle or ls: {'-', '--', '-.', ':', '', ...} or (offset, on-off-seq)
plt.grid(alpha = 0.4,color = 'red',linestyle = '-.')
plt.axis([0,7,-1.5,1.5])
```

#### 1.2 坐标轴刻度、标签、标题

```python
from matplotlib import font_manager #字体管理器
fm = font_manager.FontManager()
#获取所有字体
[font.name for font in fm.ttflist]
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.rcParams['font.family'] = 'SimHei' #字体设置
plt.rcParams['font.size'] = 25 #字体大小设置
plt.rcParams['axes.unicode_minus'] = False #负号设置

plt.figure(figsize = (9,6)) #图形尺寸设置

plt.plot(x,y)

plt.title('正弦波',pad = 20)
_ = plt.yticks([-1,0,1],labels = ['min',0,'max'])
plt.ylabel('y = sin(x)',rotation = 0,labelpad = 60)
xticks = np.arange(0,2*np.pi+0.001,np.pi/2)
_ = plt.xticks(xticks,labels = [0,r'$\frac{\pi}{2}$',r'$\pi$',r'$\frac{3\pi}{2}$',r'$2\pi$'])
```
