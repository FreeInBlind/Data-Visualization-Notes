#### 2.2 map映射

```python
import numpy as np
import pandas as pd
```

```python
df
```

```python
#df['Math']取出了一列，这一列是Series，所以map函数是Series函数
df['Math'].map({150:90,101:120}) #没有特定位置的数会变为NAN
```

```python
# 函数名任意，功能自定义
def convert(x): #x代表Math这一列的每一个元素的数值
    if x ==150:
        return 90
    elif x == 101:
        return 120
    else:
        return x
df['Math'].map(convert)
```

```python
df['Math'] = df['Math'].map(convert)
```

```python
df
```

```python
df['Python'].map(lambda x : 88 if x == 150 else 149)
```

#### 2.3 apply元素改变

```python
#apply的对象是df中的一列Series
df['En'].apply(lambda x : 100 if x < 100 else x)
```

```python
df.apply(lambda x : x.median()) #apply的对象是df中的每一列（即其本身）
```

```python
def convert(x):
    return(x.mean(),x.median(),x.count())
df.apply(convert,axis = 0) #axis为0计算列的结果，为1计算行的结果
```

```python
df.apply(lambda x: x+100)
```

#### 2.4 transform数据改变

```python
#1.一列执行多项计算
df['Python'].transform([np.sqrt,np.exp]) #Series处理
```

```python
def convert(x):
    if x.mean() > 5:
        x *= 10
    else:
        x *= -10
    return x
```

```python
#2.多项执行不同计算
df.transform({'Python':convert,'Math':np.sqrt,'En':np.sin}) #DataFrame处理
```

### 3 分箱操作

```python
df = pd.DataFrame(data = np.random.randint(0,150,size = (100,3)),
                  columns = ['Python','Tensorflow','Keras'])
df
```

```python
#1.等距分箱
pd.cut(df.Python,bins = 3)
```

```python
#指定距离分箱
pd.cut(df.Keras, #分箱操作
       bins = [0,60,90,120,150], #分箱短点
       right = False, #左闭右开
       labels = ['不及格','中等','良好','优秀']) #分箱后分类
```

```python
#2.等频分箱
pd.qcut(df.Python,q = 4, #四等分
        labels = ['差','中','良','优']) #分箱后分类
```
