#### 1.3 IO操作

```python
arr1 = np.random.random(size = (3,5))
arr1
```

```python
# 文件后缀，默认是npz
np.save('./arr1',arr1)
```

```python
np.load('./arr1.npy')
```

```python
arr2 = np.logspace(1,10,num = 10,base = 2)
arr2
```

```python
 np.savez('./file2',arr_x = arr1,arr_y = arr2)
```

```python
np.load('./file2.npz')['arr_x']
```

```python
np.load('./file2.npz')['arr_y']
```

```python
np.savetxt('./arr1.txt',arr1,delimiter = ',')
```

```python
np.savetxt('./arr1.txt',arr1,delimiter = ',',fmt = '%.5f')
```

```python
np.savetxt('./arr1.csv',arr1,delimiter = ';',fmt = '%.5f')
```
