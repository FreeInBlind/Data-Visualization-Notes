### 1 广播机制

#### 1.1 一维广播机制

```python
[0,1,2,3]*3
```

```python
import numpy as np
#sort 排序
arr1 = np.sort(np.array([0,1,2,3]*3)).reshape(4,3) #shape(4,3)
display(arr1)
arr2 = np.array([1,2,3]) #shape(3,)
display(arr2)
arr3 = arr1 + arr2 #arr2进行广播复制出四份
arr3 #jupyter会将变量输出
```

#### 1.2 二维广播机制

```python
import numpy as np
arr1 = np.sort(np.array([0,1,2,3]*3)).reshape(4,3)
display(arr1)
arr2 = np.array([[1],[2],[3],[4]])
display(arr2)
arr3 = arr1 + arr2
arr3
```

#### 1.3 三维广播机制

```python
import numpy as np
arr1 = np.array([0,1,2,3,4,5,6,7]*3).reshape(3,4,2) #shape(3,4,2)
display(arr1)
arr2 = np.array([0,1,2,3,4,5,6,7]).reshape(4,2) #shape(4,2)
display(arr2)
arr3 = arr1 + arr2
arr3
```

### 2 通用函数

#### 2.1 元素级数字函数

```python
import numpy as np
arr1 = np.array([1,4,8,9,16,25])
np.sqrt(arr1) #开方
np.square(arr1) #平方
np.clip(arr1,2,16) #取值2~16，输出array([ 2,  4,  8,  9, 16, 16])
```

```python
x = np.array([1,5,2,9,3,6,8])
y = np.array([2,4,3,7,1,9,0])
np.maximum(x,y) #返回两个数组中的比较大的值
arr2 = np.random.randint(0,10,size = (5,5))
np.inner(arr2[0],arr2) #返回一堆数组向量内积
```

#### 2.2 where函数

```python
import numpy as np
arr1 = np.array([1,3,5,7,9])
arr2 = np.array([2,4,6,8,10])
cond = np.array([True,False,True,True,False])
np.where(cond,arr1,arr2) #True选择arr1，False选择arr2的值
#输出arr([1,4,5,7,10])
```

```python
arr3 = np.random.randint(0,30,size = 20)
display(arr3)
np.where(arr3 < 15,arr3,-15)
```

#### 2.3 排序方法

```python
import numpy as np
arr = np.array([9,3,11,6,17,5,4,15,1])
display(arr)
arr.sort() #直接改变原数组
arr
```

```python
np.sort(arr) #返回深拷贝排序结果
arr = np.array([9,3,11,6,17,5,4,15,1])
index = arr.argsort() #返回从小到大排序索引 array([8, 1, 6, 5, 3, 0, 2, 7, 4])
display(arr,index)
#根据索引来排序
arr[index]
```

#### 2.4 集合运算

```python
import numpy as np

A = np.array([2, 4, 6, 8])
B = np.array([3, 4, 5, 6])

print(np.intersect1d(A, B))  # 交集 -> [4 6]
print(np.union1d(A, B))      # 并集 -> [2 3 4 5 6 8]
print(np.setdiff1d(A, B))    # 差集 (A - B) -> [2 8]
```

#### 2.5 数学和统计函数

```python
import numpy as np
arr1 = np.array([1,7,2,19,23,0,88,11,6,11])
print(arr1.min()) #计算最小值
print(arr1.argmax()) #计算最大值的索引
print(np.argwhere(arr1 > 20)) #返回大于20的元素的索引
print(np.cumsum(arr1)) #计算累加和
```

```python
arr2 = np.random.randint(0,10,size = (4,5))
display(arr2)
print(arr2.mean(axis = 0)) #计算列的平均值
print(arr2.mean(axis = 1)) #计算行的平均值
print(np.cov(arr2,rowvar = True)) #协方差公式
print(np.corrcoef(arr2,rowvar = True)) #相关性系数
```

### 3 线性代数

#### 3.1 矩阵乘积（内积）

```python
A = np.array([[4,2,3],
              [1,3,1]]) #shape(2,3)
B = np.array([[2,7],
              [-5,-7],
              [9,3]])
np.dot(A,B) #矩阵运算A的最后一维和B的第一维必须一致
```

```python
A @ B #符号 @ 表示矩阵乘积运算
```

#### 3.2 矩阵其他计算

```python
from numpy.linalg import inv,det,eig,qr,svd
A = np.array([[1,2,3],
              [2,3,4],
              [4,5,8]]) #shape(3,3)
inv(A) #逆矩阵
```

```python
det(A)
```

```python
eig(A)
```

```python
qr(A)
```
