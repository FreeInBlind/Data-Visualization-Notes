#### 3.2 薪水处理【字符串--->数值】

```python
job['salary'].str.contains('K').sum()
```

```python
#大写转小写
s = job['salary'].str
s.lower()
```

```python
#正则表达式的应用
import re
s = '20k-25k'
re.findall(r'(\d+)k-(\d+)k',s)
```

```python
salary = job['salary'].str.lower()
salary_str = salary.str
salary_str.findall(r'(\d+)k-(\d+)k')
```

```python
salary_str.extract(r'(\d+)k-(\d+)k') #数据输出为DataFrame
```

```python
salary_str.extract(r'(\d+)k-(\d+)k').astype(np.int16).mean(axis = 1)
```

```python
job['salary'] = job['salary'].str.lower().str.extract(r'(\d+)k-(\d+)k').astype(np.int16).mean(axis = 1)
```

#### 3.3 job_detail中提取技能

```python
s = job['job_detail'].str.lower()
s.isna().sum()
```

```python
job['job_detail'] = job['job_detail'].str.lower()
job['Python'] = job['job_detail'].map(lambda x : 1 if ('python' in x) else 0)
job['SQL'] = job['job_detail'].map(lambda x : 1 if ('sql' in x) or ('hive' in x) else 0)
job['Tableau'] = job['job_detail'].map(lambda x : 1 if ('tableau' in x) else 0)
job['Excel'] = job['job_detail'].map(lambda x : 1 if ('excel' in x) else 0)
job['SPSS/SAS'] = job['job_detail'].map(lambda x : 1 if ('spss' in x) or ('sas' in x) else 0)
job.head()
```

#### 3.4 处理行业信息

```python
def convert(item):
    items = item.split(',')
    if items[0] == '移动互联网' and len(items) > 1:
        return items[1]
    else:
        return items[0]
```

```python
industry = job['industryField'].map(convert)
```
